Aproximation of PI
-----
   

```python
import random as rnd
import math

stopAt = 10000000

inCircle = 0
i =0

while i<stopAt:
    i+=1
    x=rnd.uniform(-1,1)
    y=rnd.uniform(-1,1)
    radius = x**2 + y**2
    if radius<=1:
        inCircle+=1


aprxPi=inCircle/stopAt*4
print(aprxPi)

    

    
```
