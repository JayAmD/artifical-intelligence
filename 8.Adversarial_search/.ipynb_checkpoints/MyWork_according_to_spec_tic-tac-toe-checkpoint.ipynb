# Tic-tac-toe

```python
import numpy as np
```

## Pomocné funkce

```python
# kreslí hru
def printgame(g):
    for r in game: 
        pr = ""
        for i in r:
            if i == 0:
                pr += "."
            elif i == 1:
                pr += "x"
            else:
                pr += "o"
        print(pr)
        

        
# říká kdo vyhrál 0=nikdo, 1, 2        
def whowon(g):
    
    # řádek
    if (g[0][:] == [1, 1, 1]).all() or (g[1][:] == [1, 1, 1]).all() or (g[2][:] == [1, 1, 1]).all():
        return 1
            
    if (g[0][:] == [2, 2, 2]).all() or (g[1][:] == [2, 2, 2]).all() or (g[2][:] == [2, 2, 2]).all():
        return 2
    
    # 1. sloupec
    if g[0][0] == g[1][0] == g[2][0] == 1:
        return 1
    
    if g[0][0] == g[1][0] == g[2][0] == 2:
        return 2
    
    # 2. sloupec
    if g[0][1] == g[1][1] == g[2][1] == 1:
        return 1
    
    if g[0][1] == g[1][1] == g[2][1] == 2:
        return 2
    
    
    # 3. sloupec
    if g[0][2] == g[1][2] == g[2][2] == 1:
        return 1
    
    if g[0][2] == g[1][2] == g[2][2] == 2:
        return 2
    
    
    # hlavní diagonála
    if g[0][0] == g[1][1] == g[2][2] == 1:
        return 1
    
    if g[0][0] == g[1][1] == g[2][2] == 2:
        return 2
    
    
    # hlavní anti-diagonála
    if g[0][2] == g[1][1] == g[2][0] == 1:
        return 1

    if g[0][2] == g[1][1] == g[2][0] == 2:
        return 2
       
    return 0
        

# vrací prázdná místa na šachovnici    
def emptyspots(g):
    emp = []
    for i in range(3):
        for j in range(3):
            if g[i][j] == 0:
                emp.append((i, j))
    return emp
    
```

## Herní funkce

```python
def ttt_move(game, myplayer, otherplayer):
        
    score = 0
    scores = [].copy()
    coord = (0,0)
    
    g = np.copy(game)
    emp = np.copy(emptyspots(g))
    
    depth = 9-emp.size
    
    whoWon = whowon(g)
    
    if whoWon == 1:
        return None, 10-depth
    elif whoWon == 2:
        return None, -10+depth
    
    else:
        if emp.size == 0: 
            return None, 0
        else:
            for i in emp:
                
                nextG = np.copy(g)
                #print("G: ",nextG)
                nextG[i[0], i[1]] = myplayer
                #print(nextG[i[0], i[1]] ,myplayer)
                            
                #print("nextG: ",nextG)
                
                vaste, score = ttt_move(nextG, otherplayer, myplayer)
                scores.append(score) 
                
                
                #print("i: ",i, "emp: ", emp, emp.size)
                #print("\n")
        
            if myplayer == 1:
                score = max(scores)
                idxMax = scores.index(score)
                coord = emp[idxMax]
                #g[emp[idxMax]] = 1
            else:
                score = min(scores)
                idxMin = scores.index(score)
                coord = emp[idxMin]
                #g[emp[idxMin]] = 2
                
    #print("Scores: ",scores, score)

    return coord, score
```

## Hra

```python
def play(g):
        
    for i in range(9):
        
        whoWon = whowon(g)
        
        if whoWon != 0:
            print("Won: ", whowon(g))
            return
        else:
            
            myplayer = i%2 +1
            otherplayer = i+1%2 +1

            coord, score = ttt_move(g, myplayer, otherplayer)

            g[coord[0], coord[1]] = myplayer
            printgame(g)
            print(score)
            print("\n")
        
    print("Draw")

    return
```

```python
game = np.zeros((3,3), dtype=int)


play(game)
```
