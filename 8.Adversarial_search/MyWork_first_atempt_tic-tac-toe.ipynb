# Tic-tac-toe

```python
import numpy as np
import math
```

## Pomocné funkce

```python
# kreslí hru
def printgame(g):
    for r in game: 
        pr = ""
        for i in r:
            if i == 0:
                pr += "."
            elif i == 1:
                pr += "x"
            else:
                pr += "o"
        print(pr)
        

        
# říká kdo vyhrál 0=nikdo, 1, 2        
def whowon(g):
    
    # řádek
    if (g[0][:] == [1, 1, 1]).all() or (g[1][:] == [1, 1, 1]).all() or (g[2][:] == [1, 1, 1]).all():
        return 1
            
    if (g[0][:] == [2, 2, 2]).all() or (g[1][:] == [2, 2, 2]).all() or (g[2][:] == [2, 2, 2]).all():
        return 2
    
    # 1. sloupec
    if g[0][0] == g[1][0] == g[2][0] == 1:
        return 1
    
    if g[0][0] == g[1][0] == g[2][0] == 2:
        return 2
    
    # 2. sloupec
    if g[0][1] == g[1][1] == g[2][1] == 1:
        return 1
    
    if g[0][1] == g[1][1] == g[2][1] == 2:
        return 2
    
    
    # 3. sloupec
    if g[0][2] == g[1][2] == g[2][2] == 1:
        return 1
    
    if g[0][2] == g[1][2] == g[2][2] == 2:
        return 2
    
    
    # hlavní diagonála
    if g[0][0] == g[1][1] == g[2][2] == 1:
        return 1
    
    if g[0][0] == g[1][1] == g[2][2] == 2:
        return 2
    
    
    # hlavní anti-diagonála
    if g[0][2] == g[1][1] == g[2][0] == 1:
        return 1

    if g[0][2] == g[1][1] == g[2][0] == 2:
        return 2
       
    return 0
        

# vrací prázdná místa na šachovnici    
def emptyspots(g):
    emp = []
    for i in range(3):
        for j in range(3):
            if g[i][j] == 0:
                emp.append((i, j))
    return emp
    
```

## Herní funkce

```python
def ttt_move(g, myplayer, otherplayer):
        
    scores = []
    emp = emptyspots(g)  
    
    for i in emp:
            
        nextG = np.copy(g)
        nextG[i[0], i[1]] = myplayer
        score = getScore(nextG, otherplayer, myplayer)
        scores.append(score)
            
    print("Scores: ",scores)
    
    #MinMax decision only on this level
    if myplayer == 1:
        idxMax = scores.index(max(scores))
        g[emp[idxMax]] = 1
    else:
        idxMin = scores.index(min(scores))
        g[emp[idxMin]] = 2
    
    return g
                    
    
def getScore(g, myplayer, otherplayer):
    
    score = 0
    emp = np.copy(emptyspots(g))
    depth = 9 - emp.size 

    whoWon = whowon(g)
    
    if whoWon == 1:
        return 1/math.factorial(depth) #penalization for wins deap down
    elif whoWon == 2:
        return -1/math.factorial(depth)
    
    else:

        if emp.size == 0: 
            return 0
        else:
            for i in emp:
                
                nextG = np.copy(g)
                #print("G: ",nextG)
                nextG[i[0], i[1]] = myplayer
                #print(nextG[i[0], i[1]] ,myplayer)
                            
                #print("nextG: ",nextG)

                score += getScore(nextG, otherplayer, myplayer) #counting wins to determin best branch
                #print("i: ",i, "emp: ", emp, emp.size)
                #print("Score: ",score)
                #print("\n")

            return score
    
```

## Hra

```python
def play(g, myplayer, otherplayer):
    
    for i in range(9):
        
        whoWon = whowon(g)
        
        if whoWon != 0:
            print("Won: ", whowon(g))
            return
        else:
            
            myplayer = i%2 +1
            otherplayer = i+1%2 +1

            g = ttt_move(g, myplayer, otherplayer)

            printgame(g)
            print("\n")
        
    print("Draw")

    return
```
