Right now it is possible that a element moves in one cycle more than once, because he moves to a coord, that have not been processed yet.
---

```python
import numpy as np
import matplotlib.pyplot as plt


##OPTIONS
#set number of elements of each group
groupsOf=[5000, 2500, 1000]

#set preference(WARNING in the specification the PREF is incorectly called TOL. It is not the tolerance of the agent but the preferance for same agents)
#Maybe its acctualy tolerance? -_- Play with it to see what it actually does
pref=0.1

#set how far a element can see
#wARNING do NOT set to MORE then HALF of the lower dimension of the board. E.g. xDim=7, yDim=4, max visibility=2
visibility = 1

#set dimensions of the 2D board
xDim=100
yDim=100

#set maximal number of cycles
maxCycles=1000


##constants
rng = np.random.default_rng(1564865215)  # seed

board = np.zeros((xDim,yDim), dtype = int)

neighbourIdx = np.array(np.arange(-visibility, visibility+1))

groups = np.array(groupsOf)
    
moved = -1
logMoved = []
c=0
##helpers
#initGroups
def initGroups():
    for idx, group in enumerate(groups):
        for el in np.arange(group):
            board[getSuitableCoordinates()] = idx + 1
                           
#ShowHeatMap
def heatMap():
    plt.imshow( board )
    plt.show()
    
# logMoved plot
def linPlot(log):
        fig, ax = plt.subplots()

        ax.plot(range(len(log)), log, label='Moved')
        ax.legend()


#find empty coordinates
def getSuitableCoordinates():
    suitableCoordinates = None
    while suitableCoordinates == None:
        y = rng.integers(yDim)
        x = rng.integers(xDim)
        if board[y,x] == 0:
            suitableCoordinates = y,x
    return suitableCoordinates


##main logic
#iterate through 2D array using created coordinates
def oneCycle():
    for y in np.arange(yDim):
        for x in np.arange(xDim):
            currentEl = board[y,x]
            #is element NOT empty?
            if currentEl != 0:
                countSame = 0
                countOther = 0
                #look at neighbouring elements
                for row in neighbourIdx:
                    for col in neighbourIdx:
                        if row != 0 or col != 0:
                            yLookAt = y + row
                            xLookAt = x + col
                            if yLookAt >= yDim:
                                yLookAt -=yDim
                            if xLookAt >= xDim:
                                xLookAt -=xDim
                            neighbourEl = yLookAt,xLookAt
                            if board[neighbourEl] != 0:
                                if board[neighbourEl] == currentEl:
                                    countSame+=1
                                else:
                                    countOther+=1
                #move if an element is unhappy
                if countOther !=0:
                    if countSame/(countSame + countOther) < pref or countSame==0:
                            global moved
                            moved+=1
                            suitableCoordinates = getSuitableCoordinates()
                            board[suitableCoordinates] = currentEl
                            board[y,x] = 0

                            
                            
initGroups()                            
heatMap()  

while c<maxCycles and moved!=0:
    moved = 0
    oneCycle()
    logMoved.append(moved)
    c+=1
    if c%10==0:
        print('moved LAST run:',moved,'Cycle:',c)
        heatMap()
        
print('moved LAST run:',moved,'Final cycle:',c)
heatMap()
print()
linPlot(logMoved)
            
    
```
